```python
import requests
import pandas as pd
import time

# -----------------------------
# 1. Your API credentials
# -----------------------------
API_KEY = "AIzaSyAgrgmgDBBZ0bcKRQ4ccNC9sx6kbNwuMfA"  # Replace with your Google API Key
CX = "042fb27b8d32a4934"  # Replace with your Custom Search Engine ID

# -----------------------------
# 2. Search query
# -----------------------------
query = "site:fcpeuro.com inurl:/products/bmw-*-spark-plug"

# -----------------------------
# 3. Initialize storage
# -----------------------------
all_links = []

# -----------------------------
# 4. Loop through pages (10 results per page)
#    start = 1, 11, 21, ... up to 1000 results max
# -----------------------------
for start in range(1, 101, 10):  # Adjust range as needed
    url = f"https://www.googleapis.com/customsearch/v1?q={query}&key={API_KEY}&cx={CX}&start={start}"
    resp = requests.get(url).json()

    items = resp.get("items", [])
    if not items:
        break  # no more results

    for item in items:
        # Append "#fitment" to the end of each URL
        link_with_fitment = item["link"] + "#fitment"
        all_links.append(link_with_fitment)

    time.sleep(1)  # polite pause between requests

# -----------------------------
# 5. Save to CSV
# -----------------------------
df = pd.DataFrame(all_links, columns=["Product_URL"])
df.to_csv("fcpeuro_bmw-spark-plug-kit_url2.csv", index=False)

print(f"Collected {len(all_links)} product URLs.")
print("Saved to fcpeuro_bmw-spark-plug-kit_url2.csv")
```
