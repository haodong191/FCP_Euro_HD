```python
#works but shows an error msg prior to loading the scrapped data
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

def init_driver(headless=True):
    options = webdriver.ChromeOptions()
    if headless:
        options.add_argument("--headless=new")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    # options.add_argument("--disable-software-rasterizer")  # optional
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
    return driver

def scrape_fitment_table(url, driver, timeout=30):
    driver.get(url)
    wait = WebDriverWait(driver, timeout)
    try:
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "tbody.fitment-table-body")))
        # give extra time for full JS rendering
        time.sleep(2)
    except Exception as e:
        print(f"Timed out waiting for table: {e}")
        return None

    soup = BeautifulSoup(driver.page_source, "html.parser")
    tbody = soup.find("tbody", class_="fitment-table-body")
    if not tbody:
        print("Could not find the fitment table body!")
        return None

    rows = tbody.find_all("tr")
    data = []
    for row in rows:
        cells = row.find_all("td")
        cell_text = [cell.get_text(strip=True) for cell in cells]
        data.append(cell_text)

    # Determine columns based on first row
    if len(data) == 0:
        print("Table found but no rows extracted.")
        return None

    # Example: if there are 4 columns
    col_count = len(data[0])
    # You may need to adjust the names below based on actual structure
    if col_count == 4:
        columns = ["Year", "Model", "Engine", "Additional Info"]
    else:
        # generic column names
        columns = [f"col_{i+1}" for i in range(col_count)]

    df = pd.DataFrame(data, columns=columns)
    df["url"] = url
    return df

def main(urls):
    all_dfs = []
    driver = init_driver(headless=True)
    for url in urls:
        print("Scraping:", url)
        try:
            df = scrape_fitment_table(url, driver)
            if df is None:
                # retry once with visible driver
                print("Retrying with non‑headless mode...")
                driver.quit()
                driver = init_driver(headless=False)
                df = scrape_fitment_table(url, driver)
            if df is not None:
                all_dfs.append(df)
        except Exception as e:
            print(f"Unhandled error for {url}: {e}")
        finally:
            # optional: small delay between URLs
            time.sleep(1)

    driver.quit()

    if all_dfs:
        result = pd.concat(all_dfs, ignore_index=True)
        result.to_csv("fitment_all_results1.csv", index=False)
        print("Saved CSV with all results1.")
        print(result.head())
    else:
        print("No data collected.")

if __name__ == "__main__":
    urls = [
        "https://porsche.oempartsonline.com/oem-parts/porsche-inner-tie-rod-99134732200?c=bD0xJm49U2VhcmNoIFJlc3VsdHM%3D"
        # add more URLs here if desired
    ]
    main(urls)
```
